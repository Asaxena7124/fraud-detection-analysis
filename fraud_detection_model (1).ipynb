```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

sns.boxplot(x=data['Amount'])

plt.title("Outlier Detection in Amount Column")

plt.show()
```

```python
feature_cols = [f'V{i}' for i in range(1, 29)]

for col in feature_cols:
    plt.figure(figsize=(6, 4))
    sns.boxplot(x='Class', y=col, data=data)
    plt.title(f'{col} : Fraud vs Non-Fraud')
    plt.xlabel('Class (0 = Non-Fraud, 1 = Fraud)')
    plt.ylabel(col)
    plt.show()

mean_diff = {}

for col in feature_cols:
    non_fraud_mean = data[data['Class'] == 0][col].mean()
    fraud_mean = data[data['Class'] == 1][col].mean()

    diff = abs(fraud_mean - non_fraud_mean)
    mean_diff[col] = diff
mean_diff_df = pd.DataFrame(
    mean_diff.items(),
    columns=['Column', 'Mean_Difference']
)


mean_diff_df = mean_diff_df.sort_values(
    by='Mean_Difference',
    ascending=False
)

top_5 = mean_diff_df.head(5)

print("Top 5 Columns with Highest Difference Between Fraud and Non-Fraud:")
print(top_5)
plt.figure(figsize=(8, 5))
sns.barplot(
    x='Column',
    y='Mean_Difference',
    data=top_5
)

plt.title('Top 5 Features with Highest Fraud vs Non-Fraud Difference')
plt.xlabel('Features')
plt.ylabel('Mean Difference')
plt.show()
```
